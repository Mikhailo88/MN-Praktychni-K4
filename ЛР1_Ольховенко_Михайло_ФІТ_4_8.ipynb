## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:
```python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])
```

1. Обчисліть скалярний добуток двома способами: `np.dot()` і оператором `@`.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])

dot1 = np.dot(u, v)
dot2 = u @ v

print("np.dot(u, v) =", dot1)
print("u @ v =", dot2)

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

print("Норма u =", norm_u)
print("Норма v =", norm_v)

cos_angle = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

print("Косинус кута =", cos_angle)
````

```python
A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])
```

1. Переставте рядки відповідно до `row_order`.
2. Після цього переставте стовпці відповідно до `col_order`.
3. За допомогою обернених перестановок відновіть початкову матрицю.

Для побудови оберненої перестановки використайте `np.argsort()`.

````python
A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])

B = A[row_order]

print("Після перестановки рядків:")
print(B)

C = B[:, col_order]

print("\nПісля перестановки стовпців:")
print(C)

inv_row_order = np.argsort(row_order)
inv_col_order = np.argsort(col_order)

print("\nОбернена перестановка рядків:", inv_row_order)
print("Обернена перестановка стовпців:", inv_col_order)

D = C[:, inv_col_order]
D = D[inv_row_order]

print("\nВідновлена матриця:")
print(D)
````

## Завдання 15. Комутатор двох матриць

Для кожної пари матриць обчисліть комутатор
\[
[A,B] = AB - BA.
\]

Перша пара:
```python
A1 = np.array([[3, 0],
               [0, 7]])
B1 = np.array([[2, 0],
               [0, 5]])
```

Друга пара:
```python
A2 = np.array([[0, 2],
               [-1, 0]])
B2 = np.array([[4, 0],
               [0, 1]])
```

Визначте, для якої пари виконується `AB = BA`. Для перевірки використайте `np.allclose()`.

````python
A1 = np.array([
    [3, 0],
    [0, 7]
])

B1 = np.array([
    [2, 0],
    [0, 5]
])

A2 = np.array([
    [0, 2],
    [-1, 0]
])

B2 = np.array([
    [4, 0],
    [0, 1]
])

comm1 = A1 @ B1 - B1 @ A1
comm2 = A2 @ B2 - B2 @ A2

print("Комутатор першої пари [A1, B1]:")
print(comm1)

print("\nКомутатор другої пари [A2, B2]:")
print(comm2)

# Перевірка AB = BA
equal1 = np.allclose(A1 @ B1, B1 @ A1)
equal2 = np.allclose(A2 @ B2, B2 @ A2)

print("\nA1B1 = B1A1?", equal1)
print("A2B2 = B2A2?", equal2)
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

X = np.column_stack((np.ones(len(x)), x))

coefficients, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)
a, b = coefficients

y_pred = X @ coefficients

sum_squared_errors = np.sum((y - y_pred) ** 2)

print("Матриця ознак:")
print(X)

print("\nКоефіцієнт a =", a)
print("Коефіцієнт b =", b)

print("\nПрогнозні значення y:")
print(y_pred)

print("\nСума квадратів залишків =", sum_squared_errors)
````
